```python
with open('csv_20260206_08a514.txt', 'r') as f:
  data = f.readlines()
def proverka(data):
  if data.isdigit() or (data[0] == '-' and data[1:].isdigit()):
    return 'int'
  if data.count('.') == 1:
    tmp = data.replace('.', '', 1)
    if tmp.isdigit() or (tmp[0] == '-' and tmp[1:].isdigit()):
      return 'float'
  if not data.isdigit():
    return 'str'
def convert_value(value):
  if proverka(value) == 'int':
    return int(value)
  if proverka(value) == 'float':
    return float(value)
  if proverka(value) == 'str':
    return str(value)

header = [col for col in data[0].split(',')]

a = []
for i in range(1, len(data)):
  row = [col for col in data[i].split(',')]
  converted_row = [convert_value(val) for val in row]
  a.append(converted_row)

types = []
for val in a[0]:
  if isinstance(val, int):
      types.append('int')
  elif isinstance(val, float):
      types.append('float')
  else:
      types.append('str')

result = {'header': header, 'data': a, 'types': types}
print(result)



```
